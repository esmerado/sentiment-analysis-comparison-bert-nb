# Proyecto 2: Procesar y estructurar textos clínicos con modelos avanzados

## 1. Tareas principales
* **Tokenización:** Dividir el informe de unidades con sentido médico.
* **NER (Reconocimiento de Entidades Nombradas)**: Identificar automáticamente síntomas, enfermedades y tratamientos.
* **Limpieza y procesamiento:** Uso de SpaCy y NLTK para normalizar el lenguaje médico.
* **Visualización:** Mostrar de forma clara la información relevante extraída.

```python
import spacy
import pandas as pd
from spacy import displacy
import IPython
from IPython.display import display, HTML

# 1. Carga del modelo en español
nlp = spacy.load('es_core_news_sm')
```

### Tokenización

El sistema realiza la tokenización de los informes clínicos dividiendo el texto en unidades lingüísticas con significado médico, tales como palabras clave, signos clínicos y términos farmacológicos. Este paso es fundamental para permitir el análisis posterior del lenguaje natural, ya que convierte el texto libre en elementos procesables por el modelo.


```python
raw_report = """
Paciente masculino presenta cefalea intensa y náuseas. 
Se observa hipertensión arterial. Se receta Enalapril 20mg.
El paciente debe volver a consulta en 15 días.
"""

doc = nlp(raw_report)

print(f"{'Token':<15} | {'Categoría Gramatical':<20}")
print("-" * 40)
for token in doc[:10]: # Mostramos los primeros 10 tokens
    print(f"{token.text:<15} | {token.pos_:<20}")
```

### Reconocimiento de Entidades Nombradas (NER)

Se implementó un sistema de Reconocimiento de Entidades Nombradas (NER) para identificar automáticamente conceptos clínicos relevantes dentro de los informes médicos. El modelo detecta entidades pertenecientes a las categorías de síntomas, enfermedades y tratamientos, permitiendo transformar texto no estructurado en información clínica estructurada y reutilizable.


```python
extracted_entities = []

for ent in doc.ents:
    extracted_entities.append({
        "Texto": ent.text,
        "Etiqueta": ent.label_,
        "Descripción": spacy.explain(ent.label_)
    })

df_clinical = pd.DataFrame(extracted_entities)
print("Información Estructurada del Informe:")
display(df_clinical)
```

### Limpieza y Procesamiento de Datos Médicos.

Antes de la extracción de entidades, los informes clínicos son sometidos a un proceso de limpieza y normalización del lenguaje médico. Para ello se emplean herramientas como SpaCy y NLTK, las cuales permiten eliminar ruido textual, unificar formatos y preparar el texto para un análisis consistente. Este preprocesamiento mejora la precisión del reconocimiento de entidades y la robustez del sistema.


```python
cleaned_tokens = [token.lemma_ for token in doc if not token.is_stop and not token.is_punct]

print("Tokens médicos normalizados (Lematizados):")
print(cleaned_tokens)
```

### Análisis de Patrones y Visualización

La información clínica extraída es presentada de forma visual y estructurada mediante herramientas de visualización como displacy. Esta representación facilita la interpretación rápida de los datos relevantes por parte del personal médico, permitiendo identificar síntomas, diagnósticos y tratamientos directamente sobre el texto original del informe clínico.


```python
print("Visualización del Pipeline de Procesamiento Médico:")
html = displacy.render(doc, style="ent", jupyter=False) 
display(HTML(html))
```

## Resultados Obtenidos

El sistema de procesamiento de lenguaje natural logró identificar de manera correcta las entidades clínicas relevantes presentes en los informes médicos analizados. A través del pipeline automatizado, se extrajeron síntomas, enfermedades y tratamientos directamente desde texto clínico no estructurado, demostrando la capacidad del modelo para transformar información narrativa en datos estructurados.

La visualización de resultados mediante displacy permitió validar de forma inmediata la precisión del reconocimiento de entidades, facilitando la revisión y verificación de los datos extraídos. En particular, se observó una correcta identificación de términos médicos clave, lo que confirma la efectividad del enfoque híbrido basado en modelos y reglas personalizadas.
